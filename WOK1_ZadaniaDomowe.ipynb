## Zadanie 2: Własności iloczynu wewnętrznego
### Opis: Udowodnij analitycznie podstawowe własności iloczynu wewnętrznego w przestrzeniach zespolonych. Następnie zweryfikuj je numerycznie dla losowych wektorów.

✍️ Obliczenia analityczne
Udowodnij:

+ $\braket{u|\alpha v + \beta w} = \alpha\braket{u|v} + \beta\braket{u|w}$
+ $\braket{\alpha u + \beta v|w} = \alpha^{*}\braket{u|w} + \beta^{*}\braket{v|w}$
+ $\braket{u|v} = \braket{v|u}^{*}$

### ✍️ Rozwiązanie analityczne

**Definicja.** Dla $\ket{u}, \ket{v} \in \mathbb{C}^n$:
$$\braket{u|v} = \sum_{k=1}^{n} u_k^{*}\, v_k$$

Korzystamy z własności sprzężenia (Zadanie 1):
- (A) $(z_1 + z_2)^* = z_1^* + z_2^*$
- (B) $(z_1 z_2)^* = z_1^*\, z_2^*$
- (C) $(z^*)^* = z$

---

#### Własność 1: liniowość względem prawego argumentu

$k$-ta współrzędna wektora $\alpha\ket{v} + \beta\ket{w}$ to $\alpha v_k + \beta w_k$, zatem:
$$\braket{u|\alpha v + \beta w} = \sum_k u_k^{*}\,(\alpha v_k + \beta w_k) = \sum_k \big(\alpha\, u_k^{*} v_k + \beta\, u_k^{*} w_k\big)$$
$$= \alpha \sum_k u_k^{*} v_k + \beta \sum_k u_k^{*} w_k = \alpha\braket{u|v} + \beta\braket{u|w} \qquad \blacksquare$$

---

#### Własność 2: antyliniowość względem lewego argumentu

Kombinacja stoi po lewej stronie, więc jej współrzędne są sprzężone:
$$\braket{\alpha u + \beta v|w} = \sum_k (\alpha u_k + \beta v_k)^{*}\, w_k$$

Z (A) i (B):
$$(\alpha u_k + \beta v_k)^{*} = (\alpha u_k)^{*} + (\beta v_k)^{*} = \alpha^{*} u_k^{*} + \beta^{*} v_k^{*}$$

Zatem:
$$\braket{\alpha u + \beta v|w} = \sum_k \big(\alpha^{*} u_k^{*} w_k + \beta^{*} v_k^{*} w_k\big) = \alpha^{*}\sum_k u_k^{*} w_k + \beta^{*}\sum_k v_k^{*} w_k = \alpha^{*}\braket{u|w} + \beta^{*}\braket{v|w} \qquad \blacksquare$$

---

#### Własność 3: symetria sprzężona

$$\braket{v|u}^{*} = \Big(\sum_k v_k^{*}\, u_k\Big)^{*} \overset{(A),(B)}{=} \sum_k \big(v_k^{*}\big)^{*}\, u_k^{*} \overset{(C)}{=} \sum_k v_k\, u_k^{*} = \sum_k u_k^{*}\, v_k = \braket{u|v} \qquad \blacksquare$$

**Wniosek:** dla $u = v$ mamy $\braket{v|v} = \braket{v|v}^{*}$, więc $\braket{v|v} \in \mathbb{R}$ (a nawet $\braket{v|v} = \sum_k |v_k|^2 \geq 0$), co uzasadnia definicję normy.

**Uwaga:** własność 2 wynika też z własności 1 i 3:
$$\braket{\alpha u + \beta v|w} \overset{(3)}{=} \braket{w|\alpha u + \beta v}^{*} \overset{(1)}{=} \big(\alpha\braket{w|u} + \beta\braket{w|v}\big)^{*} = \alpha^{*}\braket{w|u}^{*} + \beta^{*}\braket{w|v}^{*} \overset{(3)}{=} \alpha^{*}\braket{u|w} + \beta^{*}\braket{v|w}$$

```python
# Zadanie 2 – weryfikacja numeryczna dla losowych wektorów zespolonych
rng = np.random.default_rng(42)

def losowy_wektor(n=3):
    return rng.normal(size=n) + 1j * rng.normal(size=n)

u, v, w = losowy_wektor(), losowy_wektor(), losowy_wektor()
alpha, beta = 2 - 1j, 0.5 + 3j

# Własność 1: liniowość w prawym argumencie
L1 = np.vdot(u, alpha*v + beta*w)
P1 = alpha*np.vdot(u, v) + beta*np.vdot(u, w)
print("Własność 1:", np.isclose(L1, P1), L1, P1)

# Własność 2: antyliniowość w lewym argumencie
L2 = np.vdot(alpha*u + beta*v, w)
P2 = np.conj(alpha)*np.vdot(u, w) + np.conj(beta)*np.vdot(v, w)
print("Własność 2:", np.isclose(L2, P2), L2, P2)

# Kontrprzykład: bez sprzężenia α, β równość NIE zachodzi
P2_zle = alpha*np.vdot(u, w) + beta*np.vdot(v, w)
print("Własność 2 bez sprzężeń:", np.isclose(L2, P2_zle))

# Własność 3: symetria sprzężona
L3 = np.vdot(u, v)
P3 = np.conj(np.vdot(v, u))
print("Własność 3:", np.isclose(L3, P3), L3, P3)
```
